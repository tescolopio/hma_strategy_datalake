```python
# Feature correlations with success
import polars as pl
feature_cols = ["wind_speed", "temp_c", "wind_favorable", "is_ambush",
                "is_still_hunt", "is_stalking", "is_calling"]
for col in feature_cols:
    corr = gold.select(pl.corr(col, "success")).item()
    print(f"  {col:18s}: {corr:+.3f}")
```
