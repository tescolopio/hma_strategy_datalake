# 01 — Synthetic Data Generation

Generates 5 000 realistic hunt-log records and writes them to
`data/raw/hunt_logs.csv`.  
This simulates raw telemetry that would be ingested from the
Hunt Master Academy mobile app.


```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path().resolve().parent))

from src.generate_synthetic_data import generate, OUTPUT_PATH
from pathlib import Path

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
df = generate(n=5_000)
df.write_csv(OUTPUT_PATH)
print(f"✅  {len(df)} records saved to {OUTPUT_PATH}")
```
